### Train Models

This the model training step. 

If you build data from local repository, the input files is in default **dg_cache** folder and do not need to specify.

If you download data from the links, use the data in Realistic folders.

```python
%%bash
#ML_models
projects=("FFmpeg", "linux")
ml_models=("lapredict", "tlel", "lr")

for project in "${projects[@]}"; do 
    for model in "${ml_models[@]}"; do
        echo "Training: $project - $model"
        vulguard training \
            -model $model \
            -repo_name $project \
            -repo_language C \
            -dg_save_folder . \
            -train_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-features-train.jsonl \
            -val_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-features-val.jsonl \
            -dictionary <DOWNLOAD_DIR>/data/$project/Realistic/dict-$project.jsonl
    done
done
```

```python
%%bash
#DL_models
projects=("FFmpeg", "linux")
dl_models=("deepjit", "simcom", "jitfine")

for project in "${projects[@]}"; do 
    for model in "${dl_models[@]}"; do
        echo "Training: $project - $model"
        vulguard training \
            -model $model \
            -repo_name $project \
            -repo_language C \
            -dg_save_folder . \
            -device cuda \
            -train_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-deepjit-train.jsonl \
            -val_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-deepjit-val.jsonl \
            -dictionary <DOWNLOAD_DIR>/data/$project/Realistic/dict-$project.jsonl
    done
done
```

```python
%%bash
#VCCFinder
projects=("FFmpeg", "linux")
model="vccfinder"

for project in "${projects[@]}"; do
    echo "Training: $project - $model"
    vulguard training \
        -model $model \
        -repo_name $project \
        -repo_language C \
        -dg_save_folder . \
        -device cuda \
        -train_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-vcc-features-train.jsonl \
        -val_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-vcc-features-val.jsonl \
        -dictionary <DOWNLOAD_DIR>/data/$project/Realistic/dict-$project.jsonl
done
```

### Test Models

This the model testing step. 

If you build data from local repository, the input files is in default **dg_cache** folder and do not need to specify.

If you download data from the links, use the data in Idealized folders.

```python
%%bash
#ML_models
projects=("FFmpeg", "linux")
ml_models=("lapredict", "tlel", "lr")

for project in "${projects[@]}"; do 
    for model in "${ml_models[@]}"; do
        echo "Testing: $project - $model"
        vulguard evaluating \
            -model $model \
            -repo_name $project \
            -repo_language C \
            -dg_save_folder . \
            -test_set <DOWNLOAD_DIR>/data/$project/Realistic/SETUP2-$project-features-test.jsonl \
            -dictionary <DOWNLOAD_DIR>/data/$project/Realistic/dict-$project.jsonl
    done
done
```
